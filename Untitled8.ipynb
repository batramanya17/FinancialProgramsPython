```python
def check_prime(num):
  if num <= 1:
    print(f" {num} is not a prime number")
  elif num == 2:
    print(f"{num} is a prime number")
  else:
    for i in range (2, num):
      if num%i==0:
        print(f" {num} is not a prime number")
        break
    else: # This 'else' is for the 'for' loop and executes if the loop completes without a break
      print(f"{num} is a prime number")

check_prime(13)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/balancesheetinfosys.csv')

# Convert the 'Year' column to datetime objects, coercing errors
df['Year'] = pd.to_datetime(df['Year'], errors='coerce', format='%d/%m/%Y')

# Drop rows where 'Year' became NaT, as these cannot be used for plotting
df.dropna(subset=['Year'], inplace=True)

# Sort by 'Total Assets' to make the bar plot more interpretable
df_sorted = df.sort_values(by="Total Assets")

# Create the bar plot
plt.figure(figsize=(12, 6)) # Set a good figure size
plt.bar(df_sorted["Total Assets"], df_sorted["TotalLiabilities"], color='skyblue')

# Add labels and title for clarity
plt.xlabel('Total Assets')
plt.ylabel('Total Liabilities')
plt.title('Total Liabilities vs. Total Assets')
plt.xticks(rotation=45) # Rotate x-axis labels if they overlap
plt.tight_layout() # Adjust layout to prevent labels from being cut off

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
df=sns.load_dataset('iris')
#regression plot
sns.regplot(x='sepal_length',y='petal_length',data=df)
plt.title('sepal_length vs petal_length')
plt.show()
#linear model plot
sns.lmplot(x='sepal_width',y='petal_width',data=df)
plt.title('seal_width vs petal_width')
plt.show()
```

```python
num = int(input("Enter a number: "))

if num > 0:
  print(f"{num} is a positive number.")
  if num % 2 == 0:
    print(f"{num} is an even number.")
  else:
    print(f"{num} is an odd number.")
elif num < 0:
  print(f"{num} is a negative number.")
else:
  print(f"{num} is zero.")
```
